```python
import os
from google.colab import drive

# 1. التأكد من ربط الدرايف
drive.mount('/content/drive/')

# 2. تحديد المسار الصحيح والمؤكد للملف في جوجل درايف
fpath = '/content/drive/MyDrive/quran-simple (1).txt'

# 3. قراءة الملف وحفظ السطور وتقسيمها
with open(fpath, encoding="utf-8") as fopen:
    lines = fopen.readlines()

# 4. معالجة الأسطر وتقسيمها
new_lines = []
for li in lines:
    li = li.strip("\n")
    li_split = li.split("|")
    if len(li_split) != 3:
        continue
    new_lines.append(li_split)

print(f"Total read lines: {len(new_lines)}")
print("Sample line 100:", new_lines[100])
```

```python
for item in new_lines[:10]:
    aya_text = item[2]
    words = aya_text.split()
    print(words)
    print("---")
```

```python
aya100 = new_lines[100][2]
aya100.split()
```

```python
all_quran_words = []
for item in new_lines:
    aya_text = item[2]
    words = aya_text.split()
    # all_quran_words.append(words)
    all_quran_words.extend(words)
    # print(words)
    # print("---")

print(len(all_quran_words))
```

```python
sorted_quran_words = sorted(all_quran_words)
print(sorted_quran_words[:100])
```

```python
unique_quran_words = list(set(sorted_quran_words))
print(unique_quran_words[:10])
```

```python
sorted_quran_words = sorted(all_quran_words)
unique_quran_words = list(set(sorted_quran_words))
unique_quran_words_sorted = sorted(unique_quran_words)
print(unique_quran_words_sorted[:10])
```

```python
for item in unique_quran_words_sorted[:50]:
    print(item, all_quran_words.count(item))
```

```python
from collections import Counter
count_dict = Counter(all_quran_words)
list_words_counts = list(count_dict.items())
print(list_words_counts[:10])
```

```python
# تأكد من تشغيل الخلايا السابقة أولاً لتعريف المتغير list_words_counts
three_letter_words = [v for v in list_words_counts if len(v[0]) == 3]
three_letter_words.sort(key=lambda x: -x[1])
three_letter_words[:10]
```

```python
sorted_words_by_length = sorted(unique_quran_words, key=lambda x: len(x), reverse=True)
for w in sorted_words_by_length[:10]:
    print(w)
```
